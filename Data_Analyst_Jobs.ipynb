```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/DataAnalyst.csv')
```

```python
print(df.shape)
```

```python
df.head()
```

```python
print(df.info())
```

```python
print(df.columns)
```

```python
#dropping columns
if 'Unnamed: 0' in df.columns:
    df = df.drop(columns=['Unnamed: 0'])
else:
    print("'Unnamed: 0' column not found, it might have been dropped already.")

df.head()
print(df.shape)
```

```python
#checking for null values
df.isna().sum()
```

```python
#changing values in columns
df['Easy Apply'] = df['Easy Apply'].replace(to_replace = '-1', value = 'False')
df.head()
```

```python
#General Statistics
df.describe()
```

```python
#Value Counts for categorical columns

for col in ['Job Title', 'Location', 'Type of ownership', 'Industry', 'Sector']:
    print(df[col].value_counts().head())
```

```python
#Replacing -1 values in columns
df['Industry'] = df['Industry'].replace(to_replace='-1', value='Other')
df['Type of ownership'] = df['Type of ownership'].replace(to_replace = '-1',
                            value = 'Other')
df['Sector'] = df['Sector'].replace(to_replace = '-1', value = 'Other')
```

```python
#checking value counts again
for col in ['Job Title', 'Location', 'Type of ownership',
            'Industry', 'Sector']:
    print(df[col].value_counts().head())
```

```python
#Ratings visualizations

sns.boxplot(df, y = 'Rating')
plt.xlabel('Ratings')
plt.title('Workplace Ratings')
plt.show()
```

```python
import re
df['State'] = df['Location'].str.extract(r',\s*([A-Z]{2})')
print(df['State'])
```

```python
plt.figure(figsize = [10,6])
sns.countplot(df, x = 'State', color = 'green')
plt.xlabel('State')
plt.ylabel('Count')
plt.title('Company Locations')

#most hiring companies based in California
```

```python
import re
df['Min Salary'] = df['Salary Estimate'].str.extract(r'\$?(\d+)K').astype(float)
print(df['Min Salary'])
```

```python
df['Salary Estimate'] = df['Salary Estimate'].astype(str)
df['Max Salary'] = df['Salary Estimate'].str.extract(r'-\s*\$?(\d+)K').astype(float)
print(df['Max Salary'])
```

```python
df['Avg Salary'] = (df['Min Salary'] + df['Max Salary'])/2
print(df['Avg Salary'])
```

```python
#Avg Salary
plt.figure(figsize=(10,6))
sns.histplot(df['Avg Salary'], color = 'blue')
plt.xlabel('Average Salary (in thousands)')
plt.title('Average Salary Distribution')
plt.show()
```

```python
#Average salary and state

plt.figure(figsize = [10,6])
sns.lineplot(y = 'Avg Salary', x = 'State', color = 'blue', data = df,
             linewidth = 2, marker = 'x', markersize = 10, markerfacecolor = 'black')
plt.title('Distribution of Average Salary by State')
plt.xlabel('State')
plt.ylabel('Avg Salary')
plt.show()

#Higher average salary appears to be paid in California
```

```python
#avg salary and sector

plt.figure(figsize = [18,6])
sns.lineplot(y = 'Avg Salary', x = 'Sector', data = df, color = 'red', linewidth = 2, marker = 'o', markersize = 10, markerfacecolor = 'black')
plt.title('Distribution of Average Salary by Sector')
plt.ylabel('Average Salary')
plt.xlabel('Sector')
plt.xticks(rotation=90)
plt.show()

#Sectors like biotech, real estate show a higher average salary
```
