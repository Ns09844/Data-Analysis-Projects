```python
#replace na with unknown
df_covid['Phases'] = df_covid['Phases'].replace(np.nan, 'Unknown')
df_covid['Interventions'] = df_covid['Interventions'].replace(np.nan, 'Unknown')
```

```python
#checking phases column
df_covid['Phases'].value_counts()
```

```python
df_covid = df_covid.drop(columns = ['Acronym', 'Study Documents'])
```

```python
print(df_covid.shape)
```

```python
#status visualizations
sns.countplot(data=df_covid, y='Status', color = 'blue')

plt.title('Distribution of Trial Status')
plt.show()
```

```python
#Age Column
df_covid['Age'].value_counts()
```

```python
#Cleaning Age Group Column
def categorize_age_group(age_string):
    if pd.isna(age_string): # Handle NaN values
        return 'Unknown'
    age_string = str(age_string).lower()
    if 'child' in age_string and 'adult' in age_string and 'older adult' in age_string:
        return 'All Ages'
    elif 'child' in age_string:
        return 'Child'
    elif 'older adult' in age_string:
        return 'Older Adult'
    elif 'adult' in age_string:
        return 'Adult'
    else:
        return 'Other'

df_covid['Age Group'] = df_covid['Age'].apply(categorize_age_group)

print(df_covid['Age Group'].value_counts())
```

```python
#visualizing age groups
plt.figure(figsize=(10, 6))
sns.countplot(data=df_covid, x='Age Group', order=df_covid['Age Group'].value_counts().index, palette='crest')
plt.title('Distribution of Clinical Trials by Age Group')
plt.xlabel('Age Group')
plt.ylabel('Number of Trials')
plt.show()
```

```python
sns.countplot(data=df_covid, x='Gender', hue = 'Gender')
plt.xlabel("Gender", fontdict={'family':'sans serif', 'weight':'bold', 'fontsize' : 14})
plt.ylabel('Count', fontdict={'family':'sans serif', 'weight':'bold', 'fontsize' : 14})
plt.title('Gender Eligibility of Trials')
plt.show()
```

```python
#Date Started
df_covid['Start Date'].unique()
```

```python
df_covid['Start Date'] = pd.to_datetime(df_covid['Start Date'], errors='coerce')
df_covid['Start Year'] = df_covid['Start Date'].dt.year

print(df_covid[['Start Date', 'Start Year']].head())
```

```python
def categorize_start_year(year):
    if pd.isna(year):
        return np.nan
    elif year == 2017: return '2017'
    elif year == 2018: return '2018'
    elif year == 2019: return '2019'
    elif year == 2020: return '2020'
    elif year == 2021: return '2021'
    else: return 'Other Year'

df_covid['Start Year Category'] = df_covid['Start Year'].apply(categorize_start_year)

print(df_covid['Start Year Category'].value_counts())
```

```python
#visualizing start dates of trials
plt.figure(figsize=(9,5))
sns.countplot(df_covid, x = 'Start Year Category', color = 'green')
plt.xlabel('Start Date')
plt.ylabel('Count')
plt.title('Trial Start Year')
plt.show()
```

```python
#Clinical Trial Phases
plt.figure(figsize = (10,6))
sns.countplot(data = df_covid, y = 'Phases', palette = 'coolwarm')

plt.xlabel('Count', fontdict = {'family':'sans serif', 'weight':'bold'})
plt.ylabel('Phases', fontdict = {'family':'sans serif', 'weight':'bold'})
plt.show

#conclusions: for the trials that have phases, maximum are in phase 2
```

```python
# Study Types

df_covid['Study Type'].value_counts()

#more intervention studies - trials where participants are given some kind of intervention
```

```python
#Types of interventions

intervention_types = df_covid[df_covid['Study Type'] == 'Interventional']

top_interventions = intervention_types['Interventions'].value_counts().sort_values(ascending = True)[-5:]

plt.figure(figsize=(10, 6))
plt.barh(top_interventions.index, top_interventions.values, color = 'green')
plt.xlabel('Count')
plt.ylabel('Interventions')
plt.title('Top 5 Interventions Given')
plt.tight_layout()
plt.show()
```

```python
#Top 10 sponsors or collaborators

sponsors = df_covid['Sponsor/Collaborators'].value_counts().head(10)

sns.barplot(y = sponsors.index, x = sponsors.values, palette = 'coolwarm')
plt.xlabel('Count')
plt.ylabel('Top Trial Sponsors')
plt.title('Top 10 Sponsors of COVID 19 Trials')
```
